Q2 Consider the equation $π^7$ = $e^x$.  
(a) Solve the equation by hand to get a mathematical expression for x   
(b) Use the Python to evaluate this expression for x. (from numpy import log, pi)   
(c) Test if $π^7$ and $e^x$ are equal using the relational operator ==    
(d) Evaluate $|π^7 −e^x|$. (from numpy import exp)  
(e) Estimate the rounding error in storing $π^7$ on a computer using   
• IEEE single precision arithmetic, (from numpy import finfo, float32)   
• IEEE double precision arithmetic. (from numpy import float64)  


```python
x = log(pow(pi, 7))
x
```

```python
# c) 
pow(pi, 7) == math.exp(x)
```

```python
# d) 
abs(pow(pi, 7) - math.exp(x))
```

```python
# e) 
'''
Definition (Rounding error)
The rounding error in storing x on a computer is the absolute error between the exact value of x and the stored value. Estimating rounding 
error ε is the relative machine precision on a computer ε|x| gives an estimate (or a bound) of the rounding error in storing x
on a computer.
'''

err_32 = np.finfo(np.float32).eps * float32(pow(pi, 7))
err_64 = np.finfo(np.float64).eps * float64(pow(pi, 7))

print(err_32)
print(err_64)


```

```python
from math import pi
err_32 = np.finfo(np.float32).eps * np.float32(pi ** 7)
err_64 = np.finfo(np.float64).eps * np.float64(pi ** 7)
print(err_32)
print(err_64)
```
