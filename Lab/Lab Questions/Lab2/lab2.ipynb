(ii) $\|A\|_\infty$ by hand and using Python.

(iii) $\|A\|_2$ using Python.

(b) For the matrix $B$:

(i) How can it be efficiently calculated from the matrix $A$?

(ii) Calculate $\|B\|_1$, $\|B\|_\infty$, and $\|B\|_2$.

(iii) Calculate the largest magnitude eigenvalue $|\lambda_{\max}|$ of $B$. Does $\|B\|_2$ equal $|\lambda_{\max}|$?

````python
from numpy import inf, array
from scipy.linalg import norm, eigvals
# a)
A = array([
    [3, -2, 0, 5],
    [1, 8, 7, 0],
    [-9, 5, 1, 2],
    [3, -2, 0, 5],
    [3, -4, 0, 5]
])
# i) 
A_1 = norm(A, 1)

# ii)
A_inf = norm(A, inf)

# iii)
A_2 = norm(A, 2)

print(A_1)
print(A_2)
print(A_inf)

# b)
# i) 
B = A[0:-1]

# ii) 
B_1 = norm(B, 1)
B_inf = norm(B, inf)
B_2 = norm(B, 2)

# iii)
eigs = eigvals(B)
eig_max = max(abs(eigs))
print(B_1)
print(B_2)
print(B_inf)
print(eig_max)
print(eigs)
````
