```python
# 2
import math 
from numpy import finfo

# a) 
x0 = float(3)

# b) 
def f(x):
    return (x - 1)**4 + math.e**x - 2

def f_dash(x):
    return 4 * (x - 1)**3 + math.e**x

def newtonIterate(x, g = f, g_dash = f_dash):
    return x - g(x) / g_dash(x)

x1 = newtonIterate(x0)
x2 = newtonIterate(x1)
x3 = newtonIterate(x2)
print(f"{x0}, {f(x0)}")
print(f"{x1}, {f(x1)}")
print(f"{x2}, {f(x2)}")
print(f"{x3}, {f(x3)}")

x_curr = x3
x_prev = x2
while abs(x_curr - x_prev) > finfo(float).eps:
    x_prev = x_curr
    x_curr = newtonIterate(x_prev)

xe = x_curr

invf2 = xe
print(f"{xe}, {f(xe)}")
```
